```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch  # Correctly importing as sch

# Define sequences for different species
sequences = {
    "Human": "MALWMRLLPLLALLALWGPDPAAAFVNQHLCGSHLVEALYLVCGERGFFYTPKTRREAEDLQVGQVELGGGPGAGSLQPLALEGSLQKRGIVEQCCTSICSLYQLENYCN",
    "Horse": "MGIPVGKSLLMLFTFLAFASCCIAAYRPSETLCGGELVDTLQFVCGDRGFYFSRPASRINRRSRGIVEECCFRSCDLALLETYCATPAKSERDVSTPPTVLPDDSPRYPVVKLFQYNAWKQSTQRLRRGLPALLRTRRGRMLVKELEAFREAQRHRPLIALPTEDPTPHGAAFVEVSSDLQ",
    "Rat": "MALWIRFLPLLALLILWEPRPAQAFVKQHLCGSHLVEALYLVCGERGFFYTPMSRREVEDPQVAQLELGGGPGAGDLQTLALEVARQKRGIVDQCCTSICSLYQLENYCN",
    "Chimpanzee": "MALWMRLLPLLVLLALWGPDPASAFVNQHLCGSHLVEALYLVCGERGFFYTPKTRREAEDLQVGQVELGGGPGAGSLQPLALEGSLQKRGIVEQCCTSICSLYQLENYCN",
    "Crow": "MALWIRSLPLLALLALSGPGSSHAAVSQHLCGSHLVEALYLVCGERGFFYQPKARRDVEQPLVSGPLHGELGELPFQQEEFEKVKRGIVEQCCHNTCSLYQLENYCD"
}

# Define amino acids list
amino_acids = 'ACDEFGHIKLMNPQRSTVWY'

# Calculate Amino Acid Composition (AAC) for each sequence
amnio_list = {}
for name, seq in sequences.items():
    length = len(seq)
    aac = []
    for aa in amino_acids:
        aac.append(seq.count(aa) / length * 100)  # Percentage of each amino acid
    amnio_list[name] = aac   

# Display AAC for each species
for name, seq in amnio_list.items():
    print(f"{name} : {seq}")

# Convert AAC dictionary to NumPy array
aac_array = np.array(list(amnio_list.values()))
labels = list(amnio_list.keys())

# Perform hierarchical clustering using Ward's method
linkage_matrix = sch.linkage(aac_array, method="ward")
dendrogram = sch.dendrogram(linkage_matrix, labels=labels)

# Plot the dendrogram
plt.title("Insulin Amino Acid Composition (AAC) Clustering")
plt.xlabel("Species")
plt.ylabel("Euclidean Distance")
plt.show()
```
